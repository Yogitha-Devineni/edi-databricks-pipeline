```sql
SELECT *
FROM workspace.default.edi_purchase_orders;
```

```sql
SELECT
    po_number,
    SUM(line_total) AS total_po_value
FROM workspace.default.edi_purchase_orders
GROUP BY po_number;
```

```sql
SELECT
    product_id,
    SUM(quantity) AS total_quantity
FROM workspace.default.edi_purchase_orders
GROUP BY product_id
ORDER BY total_quantity DESC;
```

```sql
SELECT
    product_id,
    SUM(line_total) AS total_value
FROM workspace.default.edi_purchase_orders
GROUP BY product_id
ORDER BY total_value DESC;
```

```sql
SELECT
    product_id,
    AVG(unit_price) AS average_unit_price
FROM workspace.default.edi_purchase_orders
GROUP BY product_id
ORDER BY average_unit_price DESC;
```

```sql
SELECT
    po_number,
    po_date,
    buyer_name,
    ship_to_name,
    department,
    COUNT(DISTINCT product_id) AS number_of_products,
    SUM(quantity) AS total_quantity,
    SUM(line_total) AS total_po_value
FROM workspace.default.edi_purchase_orders
GROUP BY
    po_number,
    po_date,
    buyer_name,
    ship_to_name,
    department;
```
